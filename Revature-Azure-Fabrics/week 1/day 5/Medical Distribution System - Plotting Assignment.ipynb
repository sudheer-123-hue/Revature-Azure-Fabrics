```
import matplotlib.pyplot as plt

Month=["Jan","Feb","Mar","Apr","May","Jun"]
Medicine_Distributed=[1200,1500,1800,1700,2100,2500]
Hospital_Orders=[30,45,50,48,60,75]

plt.plot(Month, Medicine_Distributed, marker="o", label="Medicine Distributed")
plt.plot(Month, Hospital_Orders, marker="*", label="Hospital Orders")

plt.xlabel("Month")
plt.ylabel("Values")
plt.title("Medicine Distributed vs Hospital Orders")
plt.legend()
plt.savefig("graph3.png")   # Save image
plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np

Months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
Medicine_Distributed = [500, 700, 900, 850, 1000, 1200]
Hospital_Orders = [450, 650, 800, 900, 950, 1100]

x = np.arange(len(Months))
width = 0.35

plt.bar(x - width/2, Medicine_Distributed, width)
plt.bar(x + width/2, Hospital_Orders, width)

plt.xticks(x, Months)
plt.savefig("graph bar 3.png")   # Save image
plt.show()
```
